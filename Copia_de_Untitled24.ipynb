```python
# 1. Preparar el sistema base
!sudo add-apt-repository ppa:ondrej/php -y > /dev/null 2>&1
!sudo apt-get update > /dev/null 2>&1
!sudo apt-get install mariadb-server php8.2 php8.2-cli php8.2-mysql php8.2-zip php8.2-gd php8.2-mbstring php8.2-curl php8.2-xml php8.2-bcmath unzip -y > /dev/null 2>&1
!curl -fsSL https://deb.nodesource.com/setup_20.x | sudo -E bash - > /dev/null 2>&1
!sudo apt-get install -y nodejs > /dev/null 2>&1
!curl -sS https://getcomposer.org/installer -o /tmp/composer-setup.php > /dev/null 2>&1
!php /tmp/composer-setup.php --install-dir=/usr/local/bin --filename=composer > /dev/null 2>&1

# 2. Configurar la base de datos (Arranque forzado a prueba de fallos)
!mkdir -p /var/run/mysqld
!chown mysql:mysql /var/run/mysqld
!/etc/init.d/mariadb start || /etc/init.d/mysql start
!sleep 2
!mysql -u root -e "CREATE DATABASE IF NOT EXISTS santrix_db;"
!mysql -u root -e "CREATE USER IF NOT EXISTS 'alumno'@'localhost' IDENTIFIED BY 'machinelearning';"
!mysql -u root -e "GRANT ALL PRIVILEGES ON *.* TO 'alumno'@'localhost';"
!mysql -u root -e "FLUSH PRIVILEGES;"
!mysql -u root santrix_db < /content/database.sql

# 3. Limpiar restos fallidos, clonar y configurar
!rm -rf /content/LIQUIDACION
!git clone https://github.com/DARUCLA/LIQUIDACION.git
%cd /content/LIQUIDACION
!cp .env.example .env

# 4. Inyectar credenciales línea por línea (Sin errores de sintaxis)
!echo "" >> .env
!echo "DB_CONNECTION=mysql" >> .env
!echo "DB_HOST=127.0.0.1" >> .env
!echo "DB_PORT=3306" >> .env
!echo "DB_DATABASE=santrix_db" >> .env
!echo "DB_USERNAME=alumno" >> .env
!echo "DB_PASSWORD=machinelearning" >> .env

# 5. Instalar dependencias y compilar
!composer install
!npm install
!npm run build
!php artisan key:generate
!rm -f bootstrap/cache/*.php
!php artisan config:clear
!php artisan cache:clear

print("¡Entorno reconstruido y libre de errores!")
```
